## 2️⃣ Load Data

We will load the raw jobs data from a JSON file into a Python object.  

- Using Python's built-in `json` library, we open the file in read mode (`"r"`) with UTF-8 encoding.  
- The `json.load()` function reads the JSON data and converts it into a Python dictionary or list, depending on the structure.  

```python
with open("../data/raw/raw_jobs.json", "r", encoding="utf-8") as f:
    jobs = json.load(f)
```

# CLEANING STARTS

```python
#SINCE ALL SALARY COLUMNS ARE EMPTY DROP THEM
df = df.drop('salary_avg',axis=1)
df = df.drop('salary_min',axis=1)
df = df.drop('salary_max',axis=1)
```

```python
df.info()
```

```python
#DATE POSTED TO DATETIME
# Convert date_posted to datetime
df['date_posted'] = pd.to_datetime(df['date_posted'], errors='coerce')

# Check result
df.info()
df['date_posted'].head()
```

```python
#Job_type might contain N/A values
df['job_type'].value_counts()
```

```python
df.drop('job_type',axis=1,inplace=True) 
```

```python
df.info()
```

```python
#Clean Skills Required 
print(type(df['skills_required'][0]))
```

```python
# Clean and standardize skills_required

df['skills_required'] = df['skills_required'].apply(
    lambda skills: list(
        set(
            skill.strip().lower() 
            for skill in skills
        )
    )
)

# Check result
df['skills_required'].head()
```

```python
#location
df['location'].head(10)
```

```python
df['location'].nunique()
```

```python
df['location'].value_counts().head(10)
```

```python
# Clean and standardize location column

df['location'] = df['location'].apply(
    lambda x: x.strip().title()
)

# Check result
df['location'].head(10)
```

```python
# Split location into city and state
def split_location(loc):
    parts = [p.strip() for p in loc.split(',')]
    if len(parts) == 1:
        # Only country provided
        return None, parts[0]
    else:
        # City + State
        return parts[0], parts[1]

df['city'], df['state'] = zip(*df['location'].apply(split_location))

# Check result
df[['location','city','state']].head(10)
```

```python
import numpy as np

df['state'] = df['state'].replace('India', np.nan)

# Optional: check result
df[['location','city','state']].head(10)
```

```python
# Strip extra spaces from text columns
text_cols = ['title', 'company', 'job_desc']
for col in text_cols:
    df[col] = df[col].apply(lambda x: x.strip())

# Check result
df[['title','company','job_desc']].head(5)
```

```python
df = df.drop('location', axis=1)
df.head()
```
