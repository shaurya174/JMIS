```python
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")  
# ===== Seaborn Theme (applies background, grid, and colors) =====
sns.set_theme(style="whitegrid")      # clean white grid background
sns.set_context("notebook", font_scale=1.2)  # bigger fonts for readability
sns.set_palette("Set2")               # soft, modern colors

# ===== Matplotlib rcParams Styling (titles, labels, lines) =====
plt.rcParams.update({
    'figure.figsize': (10,6),         # default figure size
    'axes.titlesize': 18,             # title font size
    'axes.labelsize': 14,             # x/y label font size
    'xtick.labelsize': 12,            # x tick font size
    'ytick.labelsize': 12,            # y tick font size
    'axes.titleweight': 'bold',       # bold titles
    'axes.labelweight': 'bold',       # bold axis labels
    'grid.color': '#dcdcdc',          # light gray grid
    'grid.linestyle': '--',           # dashed grid lines
    'grid.linewidth': 0.7,            # grid line width
    'legend.fontsize': 12,            # legend font size
    'legend.frameon': True,           # legend box
    'legend.framealpha': 0.9,         # legend transparency
    'legend.edgecolor': 'black',      # legend border color
    'lines.linewidth': 2,             # line width
    'lines.markersize': 8             # marker size
})

print("Matplotlib & Seaborn are now customized and ready to plot!")
```

```python
import pandas as pd
```

```python
skills_df = pd.read_csv('../data/processed/skill_frequency.csv')
```

```python
skills_df
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Sort skills by count descending
skills_df_sorted = skills_df.sort_values(by='Count', ascending=True)

plt.figure(figsize=(12,8))
sns.barplot(
    x='Count', 
    y='Skill', 
    data=skills_df_sorted, 
    palette='viridis'  # smooth gradient-like palette
)

plt.title("Most In-Demand Skills in Job Postings", fontsize=20, fontweight='bold')
plt.xlabel("Number of Job Postings", fontsize=14, fontweight='bold')
plt.ylabel("Skills", fontsize=14, fontweight='bold')

# Annotate bars with count
for index, value in enumerate(skills_df_sorted['Count']):
    plt.text(value + 5, index, str(value), va='center', fontsize=12)

plt.tight_layout()
plt.show()
```

```python
location_df = pd.read_csv('../data/processed/location_frequency.csv')
```

```python
location_df
```

```python
location_df_sorted = location_df.sort_values(by='Count', ascending=False).copy()

# Calculate cumulative percentage
location_df_sorted['cum_pct'] = location_df_sorted['Count'].cumsum() / location_df_sorted['Count'].sum() * 100

plt.figure(figsize=(14,8))

# Bar plot for counts
sns.barplot(x='State', y='Count', data=location_df_sorted, palette='magma')
plt.xticks(rotation=45, ha='right')

# Line plot for cumulative %
ax2 = plt.gca().twinx()
sns.lineplot(x='State', y='cum_pct', data=location_df_sorted, sort=False, color='cyan', marker='o', ax=ax2)
ax2.set_ylabel('Cumulative %', fontsize=14, fontweight='bold')
ax2.set_ylim(0, 110)

plt.title("Pareto Analysis: Job Postings by State", fontsize=20, fontweight='bold')
plt.xlabel("State", fontsize=14, fontweight='bold')
plt.ylabel("Number of Job Postings", fontsize=14, fontweight='bold')

plt.tight_layout()
plt.show()
```

```python
company_df = pd.read_csv('../data/processed/company_frequency.csv')
print(company_df)
```

```python
company_top = company_df.sort_values(by='Count', ascending=True).tail(15)  # top 15, sorted ascending

plt.figure(figsize=(14,8))
sns.barplot(
    x='Count', 
    y='Company', 
    data=company_top, 
    palette='coolwarm'
)

plt.title("Top 15 Companies by Job Postings", fontsize=20, fontweight='bold')
plt.xlabel("Number of Job Postings", fontsize=14, fontweight='bold')
plt.ylabel("Company", fontsize=14, fontweight='bold')

# Annotate bars with count
for index, value in enumerate(company_top['Count']):
    plt.text(value + 0.5, index, str(value), va='center', fontsize=12)

plt.tight_layout()
plt.show()
```

```python
weekly_df = pd.read_csv('../data/processed/weekly_skill_trends.csv')
```

```python
weekly_df
```

```python
weekly_df.info()
```

```python
weekly_df['date_posted_dt'] = pd.to_datetime(weekly_df['date_posted_dt'])

# Select top 5 skills based on total postings
skill_totals = weekly_df.drop(columns='date_posted_dt').sum().sort_values(ascending=False)
top_skills = skill_totals.head(5).index.tolist()

plt.figure(figsize=(14,7))

# Plot each top skill
for skill in top_skills:
    sns.lineplot(
        x='date_posted_dt', 
        y=skill, 
        data=weekly_df, 
        marker='o', 
        linewidth=2.5,
        label=skill
    )

plt.title("Weekly Job Demand for Top 5 Skills Over Time", fontsize=20, fontweight='bold')
plt.xlabel("Week", fontsize=14, fontweight='bold')
plt.ylabel("Number of Job Postings", fontsize=14, fontweight='bold')
plt.xticks(rotation=45)
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend(title='Skills', fontsize=12, title_fontsize=14)
plt.tight_layout()
plt.show()
```
